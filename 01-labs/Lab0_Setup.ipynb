## 6. Create your first visualization

Run the cell below. Then change the title so it includes your name and choose a different Matplotlib color. Run the cell again.

````python
# Import Matplotlib's plotting tools and give them the standard abbreviation plt.
import matplotlib.pyplot as plt

# Store the category labels and their corresponding values in two lists.
activities = ["Sleep", "Class", "Study", "Other"]
hours = [8, 3, 4, 9]

# Create a figure that contains one plotting area, called ax.
fig, ax = plt.subplots(
    figsize=(7, 4)
)  # figure will be 7 inches wide and 4 inches tall.
# Draw one bar for each activity using the matching value from hours.
ax.bar(activities, hours, color="skyblue")
# Add a descriptive title and labels for both axes.
ax.set_title("How Steve might spend a weekday")
ax.set_xlabel("Activity")
ax.set_ylabel("Hours")
# Display the completed chart in the notebook.
plt.show()
````
